# AI Code for Part A

**Prompt Used:**"Develop a python code to compute the time is required to sav efor a 25% down paymnet towards dream house based on annual income, saving percentage, house price and 4% interest rate per year."

```python
# Get user inputs for salary, savings percentage, and house cost
annual_salary = float(input("Enter your annual salary: "))
portion_saved = float(input("Enter the percent of your salary to save, as a decimal: "))
total_cost = float(input("Enter the cost of your dream home: "))

# Fixed initial values and parameters
portion_down_payment = 0.25  # 25% down payment requirement
down_payment = total_cost * portion_down_payment
current_savings = 0.0
r = 0.04  # Annual return on investment (4%)
months = 0

# Loop until savings reach the down payment target
while current_savings < down_payment:
    # Add monthly investment return
    current_savings += current_savings * (r / 12)
    # Add monthly savings from salary
    current_savings += (annual_salary / 12) * portion_saved
    # Track elapsed months
    months += 1

print("Number of months:", months)
```

# AI Code for Part B

# AI Code for Part C

**Prompt Used:**"Implement a Python code that uses besection search to determine the savings rate required to make a 250,000 down paymnet in 36 months on a house worth 1,000,000 given 7% semi-annual increments and an annual return of 4% on investments."

```python
# Get starting salary input
starting_salary = float(input("Enter the starting salary: "))

# Fixed assignment constants
total_cost = 1000000.0  # $1,000,000 home
down_payment = total_cost * 0.25  # $250,000 down payment
semi_annual_raise = 0.07  # 7% raise every 6 months
r = 0.04  # 4% annual return
target_months = 36  # Target savings period of 3 years

# Helper function to calculate total savings after 36 months for a given rate
def calculate_savings(rate_integer, salary):
    portion_saved = rate_integer / 10000.0
    current_savings = 0.0
    annual_salary = salary

    for month in range(1, target_months + 1):
        current_savings += current_savings * (r / 12)
        current_savings += (annual_salary / 12) * portion_saved
        if month % 6 == 0:
            annual_salary += annual_salary * semi_annual_raise

    return current_savings

# Verify if saving 100% (10000) is enough to reach the target within $100
if calculate_savings(10000, starting_salary) < (down_payment - 100):
    print("It is not possible to pay the down payment in three years.")
else:
    # Set search boundaries (0% to 100% mapped to 0-10000)
    low = 0
    high = 10000
    steps = 0
    found = False

    # Perform bisection search
    while low <= high:
        steps += 1
        guess = (low + high) // 2
        savings = calculate_savings(guess, starting_salary)

        # Check if savings fall within $100 of required down payment
        if abs(savings - down_payment) <= 100:
            found = True
            break
        elif savings < down_payment:
            low = guess + 1
        else:
            high = guess - 1

    if found:
        best_rate = guess / 10000.0
        print("Best savings rate:", best_rate)
        print("Steps in bisection search:", steps)
```
