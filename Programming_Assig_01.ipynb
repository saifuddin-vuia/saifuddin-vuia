# Part_A: House Hunting


```python
user_salary = float(input("Enter user annual salary: "))
save_ratio = float(input("Enter the salary saving percentage (as decimal): "))
target_home_price = float(input("Enter dream home cost: "))

required_downpay = target_home_price * 0.25
total_saved = 0.0
monthly_salary = user_salary / 12.0
r = 0.04
month_count = 0

while total_saved < required_downpay:
    total_saved += monthly_salary * save_ratio
    total_saved += total_saved * (r / 12.0)
    month_count += 1

print(f"It will take {month_count} months to save for down payment.")
```

# Part_B: Saving with a Raise

```python
user_salary = float(input("Enter user annual salary: "))
save_ratio = float(input("Enter the salary saving percentage (as decimal): "))
target_home_price = float(input("Enter dream home cost: "))
semi_annual_raise = float(input("Enter the semi-annual raise, (as decimal): "))

required_downpay = target_home_price * 0.25
total_saved = 0
monthly_salary = user_salary / 12
r = 0.04
month_count = 0

while total_saved < required_downpay:
    total_saved += total_saved * (r / 12)
    total_saved += monthly_salary * save_ratio
    month_count += 1

    if month_count % 6 == 0:
      monthly_salary += monthly_salary*semi_annual_raise

print(f"It will take {month_count} months to save for down payment.")
```

# Part_C: Finding the Right Amount to Save Away

```python
base_salary = float(input("Enter user annual salary: "))
dream_home_price = 1000000
raise_percent = .07
margin_of_error = 100
downpay_goal = dream_home_price * 0.25
target_period = 36

def calc_36m_savings(save_ratio):
    accumulated_savings = 0
    r = 0.04
    current_monthly_pay = base_salary / 12

    for current_month in range(1, target_period + 1):
        accumulated_savings += accumulated_savings * (r / 12)
        accumulated_savings += current_monthly_pay * save_ratio

        if current_month % 6 == 0:
            current_monthly_pay += current_monthly_pay*raise_percent

    return accumulated_savings

if calc_36m_savings(1.0) < (downpay_goal - margin_of_error):
    print("It is not possible to pay the down payment in three years.")

else:
      lower_rate = 0
      higher_rate = 10000
      bisection_steps = 0

      while lower_rate <= higher_rate:
          bisection_steps += 1
          guess_rate = (lower_rate+higher_rate) // 2
          decimal_rate = guess_rate / 10000
          total_saved = calc_36m_savings(decimal_rate)

          if abs(total_saved - downpay_goal) <= margin_of_error:
              break

          elif total_saved < downpay_goal:
              lower_rate = guess_rate + 1

          else:
              higher_rate = guess_rate - 1

print(f"Optimal savings rate: {guess_rate / 10000}")
print(f"Bisection search steps taken: {bisection_steps}")
```
